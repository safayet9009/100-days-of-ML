```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import pandas as pd
import time

# Set up Chrome options
options = Options()
options.add_argument("--headless")  # Run in background
options.add_argument("--disable-gpu")

# Path to chromedriver (change if needed)
driver_path = "chromedriver.exe"  # Windows: chromedriver.exe | Linux/Mac: ./chromedriver

# Set up driver
service = Service(driver_path)
driver = webdriver.Chrome(service=service, options=options)

# Prepare list to collect all data
all_data = []

# Loop through all pages
for page in range(1, 334):  # From 1 to 333
    print(f"Scraping page {page}...")
    url = f"https://www.ambitionbox.com/list-of-companies?page={page}"
    driver.get(url)
    time.sleep(3)  # Wait for JS to load

    soup = BeautifulSoup(driver.page_source, "html.parser")
    companies = soup.select("div.companyCardWrapper")

    for company in companies:
        try:
            name = company.select_one("h2.companyCardWrapper__companyName").text.strip()
        except:
            name = None
        try:
            rating = company.select_one("span.avgRating__Rating-sc-1vld6r2-1").text.strip()
        except:
            rating = None
        try:
            reviews = company.select_one("a.companyCardWrapper__ReviewsCount").text.strip()
        except:
            reviews = None
        try:
            location = company.select_one("span.companyCardWrapper__Location").text.strip()
        except:
            location = None
        try:
            industry = company.select_one("a.companyCardWrapper__CompanyType").text.strip()
        except:
            industry = None

        all_data.append({
            "Company Name": name,
            "Rating": rating,
            "Reviews": reviews,
            "Location": location,
            "Industry": industry
        })

# Close browser
driver.quit()

# Save to CSV
df = pd.DataFrame(all_data)
df.to_csv("ambitionbox_companies.csv", index=False, encoding="utf-8-sig")

print("✅ Data scraping complete. Saved to ambitionbox_companies.csv")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from bs4 import BeautifulSoup
import time
import pandas as pd

# === 1. Setup ChromeDriver Path ===
driver_path = r"C:\Path\To\chromedriver.exe"  # 🔁 Replace this with your actual path

# === 2. Set Chrome options ===
options = Options()
options.add_argument('--headless')  # comment this if you want to see the browser
options.add_argument('--disable-gpu')

# === 3. Start Chrome ===
driver = webdriver.Chrome(service=Service(driver_path), options=options)

# === 4. Load webpage ===
url = "https://www.ambitionbox.com/list-of-companies?page=1"
driver.get(url)
time.sleep(5)  # wait for JavaScript to load

# === 5. Extract HTML & close browser ===
webpage = driver.page_source
driver.quit()

# === 6. Parse with BeautifulSoup ===
soup = BeautifulSoup(webpage, "lxml")

# === 7. Find all company cards ===
companies = soup.find_all("div", class_="companyCardWrapper")

# === 8. Extract company info ===
data = []

for comp in companies:
    name = comp.select_one("h2.companyCardWrapper__companyName")
    rating = comp.select_one("span.avgRating__Rating-sc-1vld6r2-1")
    reviews = comp.select_one("a.companyCardWrapper__ReviewsCount")
    location = comp.select_one("span.companyCardWrapper__Location")
    industry = comp.select_one("a.companyCardWrapper__CompanyType")

    data.append({
        "Company Name": name.text.strip() if name else None,
        "Rating": rating.text.strip() if rating else None,
        "Reviews": reviews.text.strip() if reviews else None,
        "Location": location.text.strip() if location else None,
        "Industry": industry.text.strip() if industry else None,
    })

# === 9. Save to CSV ===
df = pd.DataFrame(data)
df.to_csv("ambitionbox_page1.csv", index=False, encoding='utf-8-sig')

print("✅ Done! Data saved to 'ambitionbox_page1.csv'")
```

```python
from selenium import webdriver
from selenium.webdriver.chrome.service import Service
from selenium.webdriver.chrome.options import Options
from webdriver_manager.chrome import ChromeDriverManager
from bs4 import BeautifulSoup
import time
import pandas as pd
from selenium.webdriver.common.by import By
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC

# === 1. Set headless Chrome options ===
options = Options()
options.add_argument('--headless')
options.add_argument('--disable-gpu')
options.add_argument('--log-level=3')

# === 2. Start WebDriver ===
driver = webdriver.Chrome(service=Service(ChromeDriverManager().install()), options=options)

# === 3. Scrape multiple pages ===
all_data = []

for page in range(1, 11):  # Pages 1 to 10
    print(f"Scraping page {page}...")
    url = f"https://www.ambitionbox.com/list-of-companies?page={page}"
    driver.get(url)

    # ✅ Wait until company cards are loaded
    try:
        WebDriverWait(driver, 15).until(
            EC.presence_of_element_located((By.CLASS_NAME, "companyCardWrapper"))
        )
    except:
        print(f"⚠️ Company cards not found on page {page}")
        continue

    soup = BeautifulSoup(driver.page_source, "html.parser")
    companies = soup.find_all("div", class_="companyCardWrapper")

    for comp in companies:
        name = comp.select_one("h2")
        rating = comp.select_one("span.average-rating")  # May vary
        reviews = comp.select_one("a.review-count")
        location = comp.select_one("span.infoEntity")  # Updated class
        industry = comp.select_one("a.industry")  # Updated class

        all_data.append({
            "Company Name": name.text.strip() if name else None,
            "Rating": rating.text.strip() if rating else None,
            "Reviews": reviews.text.strip() if reviews else None,
            "Location": location.text.strip() if location else None,
            "Industry": industry.text.strip() if industry else None,
        })

# === 4. Close browser ===
driver.quit()

# === 5. Save to CSV ===
df = pd.DataFrame(all_data)
df.to_csv("ambitionbox_companies_page1to10.csv", index=False, encoding='utf-8-sig')

print("✅ Done! Data saved to 'ambitionbox_companies_page1to10.csv'")
```
